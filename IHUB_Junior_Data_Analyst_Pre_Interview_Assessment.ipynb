## Use an appropriate statistical method to test whether the average number of required industrial training sessions is equal to 2.


```python
# Drop missing values from the column
training_sessions = talent_df['industrial_training_sessions'].dropna()

# One-sample t-test
t_stat, p_value = stats.ttest_1samp(training_sessions, popmean=2)

print(f"T-statistic: {t_stat:.4f}")
print(f"P-value: {p_value:.4f}")

# Interpretation
alpha = 0.05
if p_value < alpha:
    print("Reject the null hypothesis: The average number of trainings is significantly different from 2.")
else:
    print("Fail to reject the null hypothesis: No significant difference from 2.")
```

## Apply a suitable inferential statistics method to draw conclusions about student satisfaction with the quality of lectures received. Where 1 = Very dissatisfied, 2 = Dissatisfied, 3 = Indifferent, 4 = Satisfied, 5 = Very satisfied


```python
# Drop NaN values
satisfaction_scores = talent_df["lecture_satisfaction"].dropna()

# Run one-sample t-test against the neutral score of 3
t_stat, p_value = ttest_1samp(satisfaction_scores, popmean=3)

print(f"Mean satisfaction score: {satisfaction_scores.mean():.2f}")
print(f"T-statistic: {t_stat:.4f}")
print(f"P-value: {p_value:.4f}")

# Interpretation
alpha = 0.05
if p_value < alpha:
    print("✅ Conclusion: There is a statistically significant difference from neutral satisfaction.")
else:
    print("ℹ️ Conclusion: No significant difference from neutral satisfaction.")
```

## Use the chi-square method to analyze students' perceptions of whether the course design meets the current needs of the tech industry.where 1=strongly disagree, 2=disagree, 3=indifferent, 4=agree, and 5=strongly agree

```python
# Drop NaN values from the relevant column
course_perception = talent_df["course_relevance"].dropna()

# Get observed frequencies
observed_freq = course_perception.value_counts().sort_index()

# Expected frequencies under uniform distribution
expected_freq = [len(course_perception) / 5] * 5  # since we expect uniformity across 5 categories

# Run Chi-square test
chi2_stat, p_value = chisquare(f_obs=observed_freq, f_exp=expected_freq)

print("Observed Frequencies:\n", observed_freq)
print(f"\nChi-square Statistic: {chi2_stat:.4f}")
print(f"P-value: {p_value:.4f}")

# Interpretation
alpha = 0.05
if p_value < alpha:
    print("✅ Conclusion: The distribution is not uniform — students show significant perceptions.")
else:
    print("ℹ️ Conclusion: No significant difference from a uniform perception — responses are evenly distributed.")
```

```python
# Plot the actual distribution of responses
response_dist = course_perception.value_counts(normalize=True).sort_index() * 100

fig = px.bar(x=response_dist.index,
             y=response_dist.values,
             labels={'x': 'Perception (1=Strongly Disagree → 5=Strongly Agree)', 'y': 'Percentage'},
             title='Student Perception of Course Design Relevance')
fig.update_layout(yaxis_title="Percentage", xaxis_title="Response Scale (1-5)")
fig.show()
```

## Define entity recognition
Entity Recognition also called NER (Named Entity Recognition) is a Natural Language Processing (NLP) technique that locates and classifies named entities in text into predefined categories such as:

- Geographical Locations (example, “Nairobi”)

- Organizations (example, “World Health Organization”)

- People (example, “Bill Gates”)

- Dates (example, “January 2023”)

- Amounts (example, “$10 million”)

NER is essential for extracting structured information from unstructured text.

```python
# Imports needed for this section
import spacy
from collections import Counter
```

```python
# Load spaCy language model
nlp = spacy.load("en_core_web_sm")

# Store all named entities in a list
all_entities = []

# Loop through each column in the DataFrame
for column in text_df.columns:
    if text_df[column].dtype == 'object':  # Only process text columns
        text = " ".join(text_df[column].dropna().astype(str))
        doc = nlp(text)

        # Use doc.ents to access named entities, not tokens
        entities = [(ent.text, ent.label_, column)
                    for ent in doc.ents
                    if ent.label_ in ['GPE', 'ORG', 'PERSON', 'DATE', 'MONEY']]
        all_entities.extend(entities)

# Convert to DataFrame for further analysis or visualization
entity_df = pd.DataFrame(all_entities, columns=["Entity", "Label", "Source_Column"])
entity_df.head()
```

```python
# Imports needed for this section
from gensim import corpora, models
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
import nltk
```

```python
# Download NLTK resources
nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')
```

```python
# Combine all text columns into one list of documents
all_text = text_df.select_dtypes(include='object').fillna('').astype(str)
documents = all_text.apply(lambda row: ' '.join(row.values), axis=1)

# Preprocessing
stop_words = set(stopwords.words('english'))
tokens = [word_tokenize(doc.lower()) for doc in documents]
tokens = [[word for word in doc if word.isalpha() and word not in stop_words] for doc in tokens]

# Dictionary and Corpus
dictionary = corpora.Dictionary(tokens)
corpus = [dictionary.doc2bow(text) for text in tokens]

# Train LDA model
lda_model = models.LdaModel(corpus, num_topics=5, id2word=dictionary, passes=10, random_state=42)

# Print topics
for i, topic in lda_model.print_topics():
    print(f"Topic {i + 1}: {topic}")
```

## Using transformer models, summarize the text into one precise paragraph.

```python
# # Numpy fix
# !pip uninstall -y numpy
# !pip install numpy==1.23.5 --force-reinstall --no-cache-dir
```

```python
# Make sure you change from cpu to gpu
from transformers import pipeline

# Initialize summarization pipeline
summarizer = pipeline("summarization", model="facebook/bart-large-cnn")

# Text columns
text_columns =  ["title", "category", "story"]

# Concatenate all text entries across selected columns into one large string
all_texts = []
for col in text_columns:
    all_texts.extend(text_df[col].dropna().astype(str).tolist())

full_text = " ".join(all_texts)

# Break into smaller chunks due to model input size limits (~1024 tokens)
max_len = 1000  # slightly lower than 1024 tokens to stay safe
text_chunks = [full_text[i:i+max_len] for i in range(0, len(full_text), max_len)]

# Generate summaries
summaries = [summarizer(chunk, max_length=130, min_length=30, do_sample=False)[0]['summary_text'] for chunk in text_chunks]

# Join summaries into a final paragraph
final_summary = " ".join(summaries)

print("Summary:\n", final_summary)
```

## Apply unsupervised machine learning techniques to classify the texts into the following categories, generating ratios for each class.
Categories: Advanced Gene Editing, mRNA Technology, CAR-T Cell Therapy, Organoids and Tissue Engineering, Single-Cell Genomics, Synthetic Biology, Biological Computing, Wearable Biosensors, Microbiome Therapeutics, Nanomedicine


```python
# Text Preprocessing
import re

# Clean and combine all text columns
text_columns = text_df.select_dtypes(include='object').columns
text_df['combined_text'] = text_df[text_columns].fillna('').agg(' '.join, axis=1)

# Clean up the text
def clean_text(text):
    text = re.sub(r'\W+', ' ', text)
    return text.lower()

text_df['clean_text'] = text_df['combined_text'].apply(clean_text)
documents = text_df['clean_text'].tolist()
```

```python
# Sentence Embedding with Transformers
!pip install -q sentence-transformers
from sentence_transformers import SentenceTransformer

model = SentenceTransformer('all-MiniLM-L6-v2')  # One can also use a biomedical one like 'pritamdeka/BioBERT-mnli-snli-scinli-scitail-mednli-stsb'
document_embeddings = model.encode(documents, show_progress_bar=True)
```

```python
# Embed Category Definitions
categories = [
    "Advanced Gene Editing", "mRNA Technology", "CAR-T Cell Therapy",
    "Organoids and Tissue Engineering", "Single-Cell Genomics",
    "Synthetic Biology", "Biological Computing", "Wearable Biosensors",
    "Microbiome Therapeutics", "Nanomedicine"
]

category_embeddings = model.encode(categories)
```

```python
# Classify Each Document by Nearest Category (Cosine Similarity
from sklearn.metrics.pairwise import cosine_similarity
import numpy as np

# Compute similarities and classify
similarity_matrix = cosine_similarity(document_embeddings, category_embeddings)
predicted_indices = np.argmax(similarity_matrix, axis=1)
predicted_labels = [categories[i] for i in predicted_indices]
text_df['predicted_category'] = predicted_labels
```

```python
# Generate Category Ratios
category_counts = text_df['predicted_category'].value_counts(normalize=True) * 100
print(category_counts.round(2))
```
